```python
df_orders = cleaned_data['Orders']
df_orders['profit_margin'] = (df_orders['profit'] / df_orders['sales']) * 100
df_orders['profit_margin']
```

```python
df_orders['shipping_duration'] = (df_orders['ship_date'] - df_orders['order_date']).dt.days

df_orders['shipping_duration']
```

```python
avg_shipping_by_mode = df_orders.groupby('ship_mode')['shipping_duration'].mean()
print(avg_shipping_by_mode)
```

```python
df_orders['is_discounted'] = df_orders['discount'] > 0
df_orders['is_discounted']
```

```python
df_orders['revenue_per_unit'] = df_orders['sales'] / df_orders['quantity']
df_orders['revenue_per_unit']
```
